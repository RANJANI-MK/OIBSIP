## Handling Missing Values

The dataset contains missing values in both customer-level and transaction-level fields.

Missing values in the `age` column are replaced using the median age because age is a numerical customer attribute and the median is less sensitive to extreme values.

Rows containing missing values in `quantiy`, `price_per_unit`, `cogs`, or `total_sale` are removed because these transaction-level fields are required for accurate sales and financial analysis. Imputing these values could introduce inaccurate assumptions into the dataset.

```python
# Store the original number of rows
rows_before = len(df)

print("Rows before cleaning:", rows_before)
```

```python
# Fill missing age values with the median
df["age"] = df["age"].fillna(df["age"].median())

# Remove rows with missing transaction values
df = df.dropna(
    subset=["quantiy", "price_per_unit", "cogs", "total_sale"]
)

# Check missing values after cleaning
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Rows after missing-value cleaning:", len(df))

print("\nTotal missing values:",
      df.isnull().sum().sum())
```

## Handling Duplicate Records

Duplicate records were checked to ensure that the same transaction was not recorded more than once.

## Standardizing Inconsistent Formatting

Categorical columns were checked for inconsistent formatting such as extra spaces and inconsistent capitalization.

The `gender` and `category` columns are standardized by removing unnecessary spaces and applying consistent capitalization.

The `sale_date` column is converted from string format to a proper datetime format for accurate date-based analysis.

## Outlier Detection Using IQR

Outliers are detected using the Interquartile Range (IQR) method.

For each numerical column, the first quartile (Q1) and third quartile (Q3) are calculated.

The IQR is calculated as:

IQR = Q3 - Q1

The lower and upper boundaries are:

Lower Bound = Q1 - 1.5 × IQR
Upper Bound = Q3 + 1.5 × IQR

Values outside these boundaries are identified as potential outliers.

```python
# Numerical columns for outlier detection
numeric_cols = [
    "age",
    "quantiy",
    "price_per_unit",
    "cogs",
    "total_sale"
]

outlier_summary = []

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()
    
    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outliers
    })

outlier_df = pd.DataFrame(outlier_summary)

outlier_df
```

```python
# Calculate IQR boundaries for cogs
Q1 = df["cogs"].quantile(0.25)
Q3 = df["cogs"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Get potential cogs outliers
cogs_outliers = df[
    (df["cogs"] < lower_bound) |
    (df["cogs"] > upper_bound)
]

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of potential outliers:", len(cogs_outliers))

print("\nPotential cogs outliers:")
print(cogs_outliers[["cogs", "quantiy", "price_per_unit", "total_sale"]].head(20))
```

## Outlier Treatment Decision

The IQR method identified 87 potential outliers in the `cogs` column.

These values were inspected along with the corresponding `quantiy`, `price_per_unit`, and `total_sale` values.

The potential outliers represent higher-cost transactions and fall within the observed business ranges of the dataset. They do not appear to be invalid or data-entry errors.

Therefore, the identified `cogs` outliers are retained rather than removed or capped.

No outliers were removed from the dataset.

## Data Type Correction

The data types of the columns are corrected to ensure that each variable is stored in an appropriate format.

- Transaction and customer IDs are treated as identifiers.
- `age` and `quantiy` are converted to integer values.
- `price_per_unit`, `cogs`, and `total_sale` remain numerical values.
- `sale_date` is stored as a datetime value.
- `gender` and `category` are stored as string values.

```python
# Correct data types

df["transactions_id"] = df["transactions_id"].astype("string")
df["customer_id"] = df["customer_id"].astype("string")

df["age"] = df["age"].astype("int64")
df["quantiy"] = df["quantiy"].astype("int64")

df["gender"] = df["gender"].astype("string")
df["category"] = df["category"].astype("string")

df["price_per_unit"] = df["price_per_unit"].astype("float64")
df["cogs"] = df["cogs"].astype("float64")
df["total_sale"] = df["total_sale"].astype("float64")
```
